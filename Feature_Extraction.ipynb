```python
import librosa
import numpy as np
import pywt
from scipy.stats import skew, kurtosis
import os
import csv
from pathlib import Path
from tqdm import tqdm
import logging
import warnings

# Configuration
warnings.filterwarnings("ignore")
np.random.seed(42)
logging.basicConfig(filename='audio_extraction.log', level=logging.INFO,
                    format='%(asctime)s - %(levelname)s - %(message)s')

def safe_statistic(values, stat_name, default=0.0):
    """
    Compute a single statistic with robust error handling.

    Args:
        values (np.array): Input data
        stat_name (str): Statistic to compute ('mean', 'std', 'skew', etc.)
        default (float): Default value if computation fails

    Returns:
        float: Computed statistic or default
    """
    try:
        values = np.nan_to_num(np.asarray(values).flatten(), nan=default)
        if len(values) == 0 or not np.isfinite(values).all():
            return default
        stats = {
            'mean': np.mean,
            'std': np.std,
            'skew': lambda x: skew(x, nan_policy='omit'),
            'kurtosis': lambda x: kurtosis(x, nan_policy='omit'),
            'min': np.min,
            'max': np.max,
            'ptp': np.ptp
        }
        return stats[stat_name](values)
    except Exception as e:
        logging.warning(f"Failed to compute {stat_name}: {str(e)}")
        return default

def wavelet_decompose(signal, wavelet='db1', max_level=4):
    """
    Perform wavelet decomposition with conservative parameters.

    Args:
        signal (np.array): Input audio signal
        wavelet (str): Wavelet type
        max_level (int): Maximum decomposition levels

    Returns:
        list: Wavelet coefficients or [signal] if failed
    """
    try:
        if len(signal) < 16 or not np.isfinite(signal).all():
            logging.warning("Signal too short or invalid for decomposition")
            return [signal]
        level = min(max_level, pywt.dwt_max_level(len(signal), pywt.Wavelet(wavelet).dec_len))
        if level < 1:
            return [signal]
        coeffs = pywt.wavedec(signal, wavelet, level=level, mode='per')
        return coeffs
    except Exception as e:
        logging.error(f"Wavelet decomposition failed: {str(e)}")
        return [signal]

def extract_features(coeff, sr, prefix):
    """
    Extract robust features from a signal or subband.

    Args:
        coeff (np.array): Input coefficients or signal
        sr (int): Sample rate
        prefix (str): Feature name prefix (e.g., 'cA4', 'cD3')

    Returns:
        dict: Extracted features
    """
    features = {}
    default = 0.0
    min_length = 10

    if len(coeff) < min_length or not np.isfinite(coeff).all():
        logging.warning(f"Invalid coefficients for {prefix}")
        return {f"{prefix}_valid": 0}

    try:
        # Normalize coefficients
        coeff = np.clip(coeff, -1e6, 1e6) / (np.std(coeff) + 1e-10)

        # Time-domain features
        zcr = safe_statistic(librosa.feature.zero_crossing_rate(coeff)[0], 'mean', default)
        rms = safe_statistic(librosa.feature.rms(y=coeff)[0], 'mean', default)
        energy = safe_statistic(coeff ** 2, 'mean', default)

        # Spectral features
        n_fft = min(512, len(coeff))
        stft = np.abs(librosa.stft(coeff, n_fft=n_fft)) if len(coeff) >= n_fft else np.array([default])
        centroid = safe_statistic(librosa.feature.spectral_centroid(S=stft, sr=sr)[0], 'mean', default)
        bandwidth = safe_statistic(librosa.feature.spectral_bandwidth(S=stft, sr=sr)[0], 'mean', default)

        # MFCCs (reduced to 6 coefficients for stability)
        mfccs = np.zeros(6)
        try:
            mfccs = librosa.feature.mfcc(y=coeff, sr=sr, n_mfcc=6, n_fft=n_fft)[:, :max(1, len(coeff)//n_fft)]
        except:
            logging.warning(f"MFCC failed for {prefix}")

        # Store features
        features.update({
            f"{prefix}_ZCR_mean": zcr,
            f"{prefix}_RMS_mean": rms,
            f"{prefix}_Energy_mean": energy,
            f"{prefix}_SpectralCentroid_mean": centroid,
            f"{prefix}_SpectralBandwidth_mean": bandwidth
        })

        # MFCC statistics
        for i in range(6):
            for stat in ['mean', 'std']:
                features[f"{prefix}_MFCC_{i+1}_{stat}"] = safe_statistic(mfccs[i], stat, default)

    except Exception as e:
        logging.error(f"Feature extraction failed for {prefix}: {str(e)}")
        features[f"{prefix}_error"] = 1

    return features

def process_audio(audio_path, wavelet='db1', max_level=4):
    """
    Process a single audio file.

    Args:
        audio_path (str): Path to audio file
        wavelet (str): Wavelet type
        max_level (int): Maximum decomposition levels

    Returns:
        dict: Features or None if completely failed
    """
    try:
        # Load audio (max 30s, mono)
        signal, sr = librosa.load(audio_path, sr=22050, mono=True, duration=30.0)
        if len(signal) < sr // 20:
            logging.warning(f"Audio too short: {audio_path}")
            return None

        # Normalize signal
        signal = signal / (np.max(np.abs(signal)) + 1e-10)

        # Wavelet decomposition
        coeffs = wavelet_decompose(signal, wavelet, max_level)
        if not coeffs:
            logging.warning(f"No coefficients for {audio_path}")
            return None

        # Extract features from each subband
        features = {}
        for i, coeff in enumerate(coeffs):
            band = f"cA{max_level-i}" if i == 0 else f"cD{max_level-i+1}"
            features.update(extract_features(coeff, sr, band))

        return features if features else None
    except Exception as e:
        logging.error(f"Failed to process {audio_path}: {str(e)}")
        return None

def process_nested_folders(root_folder, output_folder):
    """
    Process all WAV files in nested folders and save CSVs to output folder.

    Args:
        root_folder (str): Root directory with WAV files in subfolders
        output_folder (str): Directory to save CSV files

    Returns:
        dict: Features by subfolder and filename
    """
    root_path = Path(root_folder)
    output_path = Path(output_folder)

    # Create output folder if it doesn't exist
    output_path.mkdir(parents=True, exist_ok=True)

    # Find all WAV files recursively
    audio_files = list(root_path.rglob('*.wav'))
    if not audio_files:
        logging.warning(f"No WAV files in {root_folder}")
        print(f"No WAV files found in {root_folder}")
        return {}

    # Group files by their parent folder
    folder_files = {}
    for audio_file in audio_files:
        parent_folder = audio_file.parent
        if parent_folder not in folder_files:
            folder_files[parent_folder] = []
        folder_files[parent_folder].append(audio_file)

    all_features = {}
    total_failed = 0
    total_processed = 0

    # Process each subfolder
    for parent_folder, files in folder_files.items():
        subfolder_name = parent_folder.name
        features_dict = {}
        failed_files = []

        print(f"Processing subfolder: {subfolder_name} ({len(files)} audio segments)")
        for audio_file in tqdm(files, desc=f"Processing {subfolder_name}"):
            features = process_audio(str(audio_file))
            if features:
                features_dict[audio_file.name] = features
                total_processed += 1
            else:
                failed_files.append(audio_file.name)

        if failed_files:
            logging.info(f"Subfolder {subfolder_name}: Failed {len(failed_files)} files: {failed_files}")
            print(f"Subfolder {subfolder_name}: Failed {len(failed_files)} files. See audio_extraction.log")
            total_failed += len(failed_files)

        # Save CSV for this subfolder
        if features_dict:
            # Sanitize subfolder name for filename
            csv_name = f"{subfolder_name.replace(' ', '_').replace('/', '_')}.csv"
            csv_path = output_path / csv_name

            try:
                columns = set()
                for features in features_dict.values():
                    columns.update(features.keys())
                columns = sorted(columns)

                with open(csv_path, 'w', newline='') as f:
                    writer = csv.DictWriter(f, fieldnames=['Filename'] + columns)
                    writer.writeheader()
                    for filename, features in features_dict.items():
                        row = {'Filename': filename}
                        row.update({col: features.get(col, 0.0) for col in columns})
                        writer.writerow(row)
                print(f"Saved CSV for {subfolder_name} to {csv_path}")
            except Exception as e:
                logging.error(f"CSV save failed for {subfolder_name}: {str(e)}")
                print(f"Failed to save CSV for {subfolder_name}: {str(e)}")

        all_features[str(parent_folder)] = features_dict

    print(f"Processed {total_processed} audio segments successfully across {len(folder_files)} patients")
    if total_failed:
        print(f"Total failed audio segments: {total_failed}. See audio_extraction.log")

    return all_features

if __name__ == "__main__":
    root_folder = "/content/drive/MyDrive/Final_Year_Project/Result/PD_Result/Chunked_Audio_PD/"
    output_folder = "/content/drive/MyDrive/Final_Year_Project/Result/PD_Result/New_PD_Result/"
    features = process_nested_folders(root_folder, output_folder)
```
