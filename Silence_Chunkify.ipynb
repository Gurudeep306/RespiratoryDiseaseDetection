```python
from pydub import AudioSegment
import webrtcvad
from pydub.playback import play
from IPython.display import display
from IPython.display import Audio
import numpy as np
import librosa
import matplotlib.pyplot as plt
from pydub.silence import split_on_silence
audio_file = "ID31_hc_0_1_1.wav"
display(Audio(audio_file, autoplay=False))
```

```python
audio = AudioSegment.from_wav(audio_file)
```

```python
audio_data, sr = librosa.load(audio_file)
```

```python
Chunks = split_on_silence(audio,
                        min_silence_len=500,
                        silence_thresh=-40
                        )
non_silent_audio = AudioSegment.empty()
for chunk in Chunks:
    non_silent_audio += chunk
```

```python
non_silent_audio.export("audio_without_silence.wav", format="wav")
res_audio = Audio("audio_without_silence.wav", autoplay=False)
display(res_audio)
```

```python
non_silent_audio = AudioSegment.from_wav("audio_without_silence.wav")
total_duration = len(non_silent_audio)
segment_duration = total_duration / 50
timestamps = [int(i * segment_duration) for i in range(51)]
for i in range(len(timestamps) - 1):
    start_time = timestamps[i]
    end_time = timestamps[i + 1]
    segment = non_silent_audio[start_time:end_time]
    export_path = f"/content/drive/MyDrive/Final_Year_Project/Result/HC_Result/New_7/Patient_18/segment_{i}.wav"
    segment.export(export_path, format="wav")

print("Audio divided into 50 segments and exported to the specified folder successfully.")
```
